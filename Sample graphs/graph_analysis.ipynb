
# Graph Analysis: Good vs. Bad Performance Graphs

This notebook displays different evaluation metric graphs and explains why they indicate good or bad model performance.

## Graph Definitions:
- **Confusion Matrix**: Shows the distribution of predicted vs actual values.
- **Normalized Confusion Matrix**: Helps visualize class imbalances and model accuracy per class.
- **F1 Score Curve**: Depicts the balance between precision and recall.
- **Precision-Recall Curve**: Indicates the tradeoff between precision and recall.
- **ROC Curve**: Measures the model's ability to distinguish between classes.
- **Results (Good/Bad)**: Compares well-performing vs. poorly performing model results.

## Interpretation of Good vs. Bad Graphs:
- **Good Graphs:** Show clear separability, high precision-recall balance, and high AUC values.
- **Bad Graphs:** Indicate poor class distinction, low AUC, and inconsistent precision-recall values.


```python
import matplotlib.pyplot as plt
import os
from IPython.display import display, Markdown

# Image files with their descriptions and explanations
image_data = [
    ("confusion_matrix.png", "Confusion Matrix", "A confusion matrix visualizes the performance of a classification model by showing the true vs. predicted labels. A good confusion matrix has high diagonal values (correct predictions) and low off-diagonal values (misclassifications)."),
    ("normalized_confusion_matrix.png", "Normalized Confusion Matrix", "Similar to a confusion matrix but normalized, making it easier to compare across datasets. A good matrix has strong values on the diagonal."),
    ("f1_score_curve.png", "F1 Score Curve", "This curve shows the F1 score across different thresholds. A good curve has a peak at a high value, indicating a strong balance between precision and recall."),
    ("precision_recall_curve.png", "Precision-Recall Curve", "This graph shows the trade-off between precision and recall. A good curve stays high (indicating strong precision and recall). A bad curve drops sharply."),
    ("r_curve.png", "Recall Curve", "Shows how recall changes with different thresholds. A good recall curve maintains high recall for a broad range of thresholds."),
    ("p_curve.png", "Precision Curve", "Displays precision variation across thresholds. A good curve maintains high precision over a wide range."),
    ("results_good.png", "Good Model Results", "This image represents well-classified examples, indicating a well-performing model."),
    ("results_bad.png", "Bad Model Results", "This image shows misclassified examples, indicating model weaknesses.")
]

# Iterate through images and display in separate cells
for img_file, title, explanation in image_data:
    display(Markdown(f"## {title}"))
    display(Markdown(f"**Definition:** {explanation}"))
    
    # Load and display image
    image = plt.imread(img_file)
    plt.figure(figsize=(6, 6))
    plt.imshow(image)
    plt.axis("off")
    plt.show()
```
